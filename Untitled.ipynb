```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r'Salary_dataset.csv')
X = df['YearsExperience']
y = df['Salary']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=.2, random_state=43)


plt.scatter(X_train, y_train)
plt.xlabel("Experience")
plt.ylabel("Salary")





```

```python
def cost_function(x,y,w,b):
    m = len(x) # number of training examples 
    cost_sum = 0 # initial cost
    for i in range(m):
        f = w *x[i] +b 
        mse = (f-y[i])**2
        cost_sum += mse
    total_cost = 1/(2*m)*cost_sum
    return total_cost
```

```python
def gradient_function(x,y,w,b):
    m = len(x)
    dc_dw = 0
    dc_db =  0
    
    for i in range(m):
        f = w * x[i] +b
        
        dc_dw += (f-y[i]) *  x[i]
        
        dc_db += (f-y[i])
    dc_dw = (1/m) * dc_dw
    dc_db = (1/m) * dc_db
    
    return dc_dw, dc_db
```

```python
def gradient_descent(x,y,alpha,iterations):
    w = 0
    b= 0
    dc_dw, dc_db = gradient_function(x,y,w,b)
    
    for i in range(iterations):
        w = w - alpha *dc_dw
        b = b - alpha * dc_db
        print(f"Iteration{i}: Cost {cost_function(x,y,w,b)}")
        
    return w, b
```

```python
learning_rate =0.000015
iterations = 1625

final_w, final_b = gradient_descent(X_train.values,y_train.values, learning_rate,iterations)

print(f'final w: {final_w}, final_b {final_b}')
```

```python
plt.scatter(X_train, y_train)

x_vals = np.linspace(min(X_train), max(X_train), 100)
y_vals = final_w* x_vals + final_b

plt.plot(x_vals,y_vals, color='red', label='regression line')
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```
